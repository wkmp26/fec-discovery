```python
import pandas as pd

# Load raw file (all as strings initially)
df = pd.read_csv("comm_pac_23_34.txt", sep="|", header=None, dtype=str)

# Define column names from your schema
columns = [
    "CMTE_ID",                 # 1. Committee ID making the disbursement
    "AMNDT_IND",               # 2. Amendment indicator
    "RPT_TP",                  # 3. Report type
    "ELECTION_TP",             # 4. Election type (e.g., P2024)
    "IMAGE_NUM",               # 5. Image number
    "TRANSACTION_TP",          # 6. Transaction type (e.g., 24K)
    "ENTITY_TP",               # 7. Entity type (e.g., CCM)
    "NAME",                    # 8. Recipient name
    "CITY",                    # 9. Recipient city
    "STATE",                   # 10. Recipient state
    "ZIP_CODE",                # 11. Recipient ZIP
    "EMPLOYER",                # 12. Employer (may be blank)
    "OCCUPATION",              # 13. Occupation (may be blank)
    "TRANSACTION_DT",          # 14. Date of disbursement
    "TRANSACTION_AMT",         # 15. Amount
    "OTHER_ID",                # 16. Other ID (if recipient is a committee)
    "CAND_ID",                 # 17. Candidate ID (if applicable)
    "TRAN_ID",                 # 18. Transaction ID
    "FILE_NUM",                # 19. File number
    "MEMO_CD",                 # 20. Memo code
    "MEMO_TEXT",               # 21. Memo text
    "SUB_ID"                   # 22. Unique record ID
]


# Assign column names
df.columns = columns


# Preview
print(df.head())
```

```python
# Assuming df is already loaded and cleaned as before
df["TRANSACTION_AMT"] = pd.to_numeric(df["TRANSACTION_AMT"], errors="coerce")

# Filter out null candidate IDs if needed
df = df[df["CAND_ID"].notna()]

# Create the interaction matrix
interaction_matrix = (
    df.pivot_table(
        index="CAND_ID",          # rows
        columns="CMTE_ID",        # columns
        values="TRANSACTION_AMT", # cell values
        aggfunc="sum",            # sum of transactions
        fill_value=0              # replace NaNs with 0
    )
)

# Preview
print(interaction_matrix.head())
```

```python
interaction_matrix.sum(axis=0).idxmax()
```

```python
import umap.umap_ as umap
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import StandardScaler

# pivot table creation, and feature engineering!!

scaler = StandardScaler()

#30
svd = TruncatedSVD(n_components=12, random_state=56)
latent = svd.fit_transform(interaction_matrix)
scaler_latent = scaler.fit_transform(latent)

scaled_latent_df = pd.DataFrame(
    scaler_latent,
    index=interaction_matrix.index,
    columns=[f"latent_{i+1}" for i in range(scaler_latent.shape[1])]
)

reducer = umap.UMAP(random_state=26)
embedding = reducer.fit_transform(scaled_latent_df)

embeddings_df = pd.DataFrame(embedding, index= scaled_latent_df.index, columns=["UMAP1","UMAP2"])
#embeddings_with_clusters = pd.merge(cluster_labels, embeddings_df, left_index=True, right_index=True)

plt.figure(figsize=(8,6))
sns.scatterplot(
    data=embeddings_df,
    x="UMAP1",y="UMAP2",
    palette='mako'
)
```
